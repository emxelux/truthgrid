```python
from langchain_tavily import TavilySearch
from dotenv import load_dotenv
load_dotenv()
```

```python
tavily = TavilySearch(
    max_results = 3
)
```

```python
response = tavily.invoke("What are the latest development in AI")
```

```python
answers = ""
for resp in response['results']:
    title = resp["title"]
    content = resp["content"]
    score = resp["score"]
    url = resp['url']
    answers += f"url: {url} \n title: {title} \n content: {content} \n score: {score} \n url: {url}"
```
